# DAPS for 3D Photon Flux Estimation (Poisson SPAD) — **Repaired inner loop (annealed PnP-ULA)**

This notebook **repairs** the DAPS sampler for the SPAD photon-flux inverse problem. The unmodified baseline lives in `DAPS_3D_Poisson_Exact.ipynb` for head-to-head comparison.

## What was wrong (estimator geometry, not the likelihood)

The Poisson likelihood here is already **exact** (§8), so likelihood mis-specification cannot be the cause of the salt-and-pepper artifact. The cause is *separability*: DAPS approximates $p(x_0\mid x_t)\approx\mathcal N\big(\hat x_0(x_t),\,\sigma_t^2 I\big)$, so the inner Langevin target $p(y\mid x_0)\,\mathcal N(\hat x_0,\sigma_t^2 I)$ factorizes over voxels — its stationary distribution is a product of per-voxel marginals. The coupled prior $p(x_0)$ survives only as the *location* of the anchor, evaluated once per outer step. At 0.001–0.005 ppp the resulting per-voxel posterior **is** the sparkle: voxels with a photon get pushed up, the rest get pulled down, and nothing in the step can move evidence sideways.

## The repair

Exact decomposition of the conditional score:
$$\nabla_{x_0}\log p(x_0\mid x_t)\;=\;\underbrace{\nabla\log p(x_0)}_{\text{coupled, network}}\;+\;\underbrace{\tfrac{\sqrt{\bar\alpha_t}}{1-\bar\alpha_t}\big(x^{\text{DDPM}}_t-\sqrt{\bar\alpha_t}\,x_0\big)}_{\text{separable, exact}}.$$
We keep the exact separable **tether** and represent $\nabla\log p(x_0)$ by the **denoiser score at a small smoothing level** $\varepsilon$ (Tweedie at $t_\varepsilon$), evaluated *inside every inner iteration*:
$$x\;\leftarrow\;x+\delta\Big[\nabla\log p(y\mid x)\;+\;\tfrac{x_t-x}{\sigma_t^2}\;+\;s_{\theta,\varepsilon}(x)\Big]+\sqrt{2\delta}\,\xi .$$
(We keep the iterate in EDM convention, where $\tfrac{x_t-x}{\sigma_t^2}$ is *identical* to the DDPM-form tether above.) Every inner iteration now applies a full, **coupled** denoiser, so a photon's evidence propagates spatially and temporally *within* each outer step — instead of the prior voting once and a separable solve running to convergence. This is annealed **PnP-ULA** (Laumont et al., "When Langevin Meets Tweedie", SIAM J. Imaging Sci. 2022), with the low-photon lineage of Melidonis et al. and Klatzer et al. Our log-flux latent $x\in[-1,1]$ keeps the Poisson gradient bounded (§8), so plain projected ULA is stable without reflection.

## Prior composition: naive averaging (for now)

The old soft blend targets $p_S^{w}p_T^{1-w}$ with $w\to0$ at fine scales — switching **off** the spatial expert exactly where impulse rejection happens (an isolated photon spike is *maximally* implausible spatially but only mildly implausible temporally). Here both experts stay on at **every** noise level via the unweighted average of their $\varepsilon$-predictions:
$$\varepsilon_{\text{comp}}=\tfrac12\big(\varepsilon_S+\varepsilon_T\big)\;\Longleftrightarrow\;s_{\text{comp}}=\tfrac12\big(s_S+s_T\big),$$
i.e. the score of the geometric mean $p_S^{1/2}p_T^{1/2}$ (a tempered product of experts). Caveat: naive score composition is biased — the product of noised marginals is not the noised product (Du et al., ICML 2023) — but the inner ULA iterations double as exactly the Langevin corrector steps that mitigate this. `PRIOR_COMPOSITION` also supports `"sum"` (full PoE; halve the step size) and `"blend"` (the old sigmoid blend, kept for A/B).

## Forward model (unchanged, exact Poisson)

$x\in\mathbb{R}^{T\times H\times W}$ is normalized log-flux in $[-1,1]$ with $\log\phi=\tfrac{L}{2}(x+1)$, $L=\log\phi_{\max}$; per voxel $N_{kij}\sim\mathrm{Poisson}\big(B_k(\alpha\,\phi_{kij}+d)\big)$. The log-likelihood and its score are voxel-wise separable **by physics** (photon detections are conditionally independent given the flux) and computed analytically (§8). Photon-count pooling is **off by default** in this notebook (`PNP_POOL_INIT = 1`) so the repair is tested cleanly; coupling now enters through the prior score inside the likelihood-facing loop, not through the forward operator.

## 3. Load 1D temporal prior

Trained on per-pixel log-flux trajectories of length $T=1024$, normalized to $[-1,1]$.  Predicts the noise $\varepsilon$ at a given DDPM timestep ($\texttt{learn\_sigma}=\texttt{False}$).

````python
def load_temporal_diffusion_model(checkpoint_path=None):
    defaults = temporal_model_and_diffusion_defaults()
    defaults.update({
        'sequence_length'      : VOL_T,
        'num_channels'         : NUM_CHANNELS_1D,
        'diffusion_steps'      : NUM_DIFFUSION_STEPS,
        'channel_mult'         : channel_mult_1d,
        'noise_schedule'       : noise_schedule_1d,
        'learn_sigma'          : False,
        'attention_resolutions': attention_resolutions_1d,
    })
    model_1d, diffusion_1d = create_temporal_model_and_diffusion(**defaults)
    if checkpoint_path and os.path.exists(checkpoint_path):
        print(f"Loading 1D checkpoint from {checkpoint_path}")
        sd = torch.load(checkpoint_path, map_location='cpu')
        model_1d.load_state_dict(sd)
    else:
        print(f"WARNING: 1D checkpoint not found at {checkpoint_path}; using random init.")
    model_1d = model_1d.to(device).eval()
    for p in model_1d.parameters():
        p.requires_grad = False
    return model_1d, diffusion_1d

model_1d, diffusion_1d = load_temporal_diffusion_model(MODEL_CHECKPOINT_1D)
print(f"1D prior params={sum(p.numel() for p in model_1d.parameters()):,}")
````

## 6. SPAD volume simulation

Per-frame Bernoulli with detection probability $1-e^{-N}$ aggregated to per-bin Binomial -- numerically equivalent to Poisson at the PPP regime we operate in (≤ 0.01).  Streamed per-frame to keep memory bounded.

````python
def generate_spad_volume(flux_vol, target_ppp=TARGET_PPP, d=DARK_COUNT,
                          n_frames=N_SPAD_FRAMES, num_bins=None, seed=0, verbose=True):
    """
    flux_vol : [T,H,W] float32, linear flux in arbitrary units.
    Returns  : bin_counts [T,H,W] float32, bin_sizes [T] float32, ppp_scale (scalar).
    """
    if num_bins is None: num_bins = flux_vol.shape[0]
    Tvol, H, W   = flux_vol.shape
    rng          = np.random.default_rng(seed)
    flux_max     = float(flux_vol.max())
    I_norm       = (flux_vol / flux_max).astype(np.float32, copy=False)
    I_mean_glob  = float(I_norm.mean())
    a_scale      = target_ppp / max(I_mean_glob, 1e-12)
    ppp_scale_3d = a_scale / flux_max

    fpb_edges      = np.linspace(0, n_frames, num_bins + 1, dtype=np.int64)
    bin_sizes      = np.diff(fpb_edges).astype(np.int64)

    if num_bins != Tvol:
        t_src = np.linspace(0.0, 1.0, Tvol,    dtype=np.float32)
        t_dst = np.linspace(0.0, 1.0, num_bins, dtype=np.float32)
        I_use = np.empty((num_bins, H, W), dtype=np.float32)
        for i in range(H):
            for j in range(W):
                I_use[:, i, j] = np.interp(t_dst, t_src, I_norm[:, i, j])
    else:
        I_use = I_norm

    bin_counts = np.empty((num_bins, H, W), dtype=np.float32)
    a32 = np.float32(a_scale); d32 = np.float32(d)
    for k in range(num_bins):
        N_k = a32 * I_use[k] + d32
        p_k = -np.expm1(-N_k)
        np.clip(p_k, 0.0, 1.0 - 1e-7, out=p_k)
        bin_counts[k] = rng.binomial(n=int(bin_sizes[k]), p=p_k).astype(np.float32, copy=False)
        if verbose and (k % 256) == 0:
            print(f"  [SPAD sim] frame {k}/{num_bins}", flush=True)
    return bin_counts, bin_sizes.astype(np.float32), float(ppp_scale_3d)
````

## 7. Prior eps-prediction and composition

Both priors take a noisy log-flux iterate in **DDPM convention** ($x^{\text{DDPM}}_t = \sqrt{\bar\alpha_t}\,x_0+\sqrt{1-\bar\alpha_t}\,\varepsilon$) and a discrete timestep $t\in\{0,\dots,999\}$, and return $\hat\varepsilon$.

* `eps_from_2d_prior` -- runs the 2D model on every frame (chunked over $T$), keeps the eps half of the `(eps, var)` output.
* `eps_from_1d_prior` -- runs the 1D model on every pixel trajectory (chunked over $H\!\cdot\!W$).
* `predict_eps_blended` -- the old sigmoid soft blend, kept only for A/B via `PRIOR_COMPOSITION="blend"`.
* `predict_eps_composed` -- composition dispatcher. Default `"avg"`: **naive averaging** $\varepsilon_{\text{comp}}=\tfrac12(\varepsilon_S+\varepsilon_T)$, equivalent to averaging the scores (geometric-mean / tempered product of experts). `"sum"` gives the full product of experts (halve the Langevin step size). Both experts are evaluated at **every** $\sigma$ — no band, no $w\to0$ hand-off to the spatially-blind 1D prior at fine scales.
* `prior_score_smoothed` -- the PnP-ULA prior score $s_{\theta,\varepsilon}(x)=\nabla\log p_\varepsilon(x)$, $p_\varepsilon=p_{\text{data}}\!\ast\!\mathcal N(0,\sigma_\varepsilon^2 I)$, computed from the composed eps-prediction at the small timestep $t_\varepsilon$: $s_{\theta,\varepsilon}(x)=-\hat\varepsilon_{\text{comp}}(\sqrt{\bar\alpha_{t_\varepsilon}}\,x,\,t_\varepsilon)/\sigma_\varepsilon$ (equivalently $(D_\varepsilon(x)-x)/\sigma_\varepsilon^2$ with the Tweedie denoiser $D_\varepsilon$). This is the coupled term applied inside every inner Langevin iteration in §10.

````python
def _amp_ctx(use_amp, on_cuda, dtype_str="fp16"):
    if not (use_amp and on_cuda):
        return _nullcontext()
    dtype = torch.float16 if dtype_str == "fp16" else torch.bfloat16
    try:
        return torch.amp.autocast("cuda", dtype=dtype)
    except (AttributeError, TypeError):
        return torch.cuda.amp.autocast(dtype=dtype)

@torch.no_grad()
def eps_from_2d_prior(x_ddpm, t_idx, model_2d, prior_info,
                       frame_chunk=FRAME_CHUNK,
                       channels_last=USE_CHANNELS_LAST,
                       use_amp=USE_AMP, amp_dtype=AMP_DTYPE):
    """Apply the 2D spatial prior to every frame of x_ddpm[B,T,H,W]; return eps[B,T,H,W]."""
    in_ch       = prior_info["in_channels"]
    learn_sigma = prior_info["learn_sigma"]
    B, Tlen, H, W = x_ddpm.shape
    frames  = x_ddpm.reshape(B * Tlen, H, W)
    eps_out = torch.empty_like(frames)
    BK = frames.shape[0]
    amp_ctx = _amp_ctx(use_amp, x_ddpm.is_cuda, amp_dtype)
    for start in range(0, BK, frame_chunk):
        end = min(start + frame_chunk, BK)
        u = frames[start:end].unsqueeze(1)                                  # [b,1,H,W]
        if in_ch != 1:
            u = u.expand(-1, in_ch, -1, -1).contiguous()
        if channels_last:
            u = u.contiguous(memory_format=torch.channels_last)
        t_batch = torch.full((u.shape[0],), int(t_idx), device=u.device, dtype=torch.long)
        with amp_ctx:
            out = model_2d(u, t_batch).float()
        eps_chs   = out[:, :in_ch] if learn_sigma else out
        eps_frame = eps_chs.mean(dim=1) if in_ch > 1 else eps_chs.squeeze(1)
        eps_out[start:end] = eps_frame
    return eps_out.reshape(B, Tlen, H, W).contiguous()

@torch.no_grad()
def eps_from_1d_prior(x_ddpm, t_idx, model_1d,
                       pixel_chunk=PIXEL_CHUNK,
                       use_amp=USE_AMP, amp_dtype=AMP_DTYPE):
    """Apply the 1D temporal prior to every pixel trajectory of x_ddpm[B,T,H,W]."""
    B, Tlen, H, W = x_ddpm.shape
    traj = x_ddpm.permute(0, 2, 3, 1).reshape(B * H * W, 1, Tlen)
    out  = torch.empty_like(traj)
    amp_ctx = _amp_ctx(use_amp, x_ddpm.is_cuda, amp_dtype)
    N = traj.shape[0]
    for start in range(0, N, pixel_chunk):
        end = min(start + pixel_chunk, N)
        u = traj[start:end]
        t_batch = torch.full((u.shape[0],), int(t_idx), device=u.device, dtype=torch.long)
        with amp_ctx:
            out[start:end] = model_1d(u, t_batch).float()
    return out.reshape(B, H, W, Tlen).permute(0, 3, 1, 2).contiguous()

def blend_weight_high(sigma_here, switch_sigma=SWITCH_SIGMA, blend_h=BLEND_H):
    """Smooth weight on the HIGH-sigma-phase prior (sigmoid in log-sigma).

    blend_h -> 0 recovers the hard switch (step at sigma == switch_sigma);
    larger blend_h widens the crossover.  Returns a scalar in (0, 1).
    """
    z = (math.log(max(float(sigma_here), 1e-12)) - math.log(switch_sigma)) / max(float(blend_h), 1e-8)
    return 1.0 / (1.0 + math.exp(-z))

@torch.no_grad()
def predict_eps_blended(x_ddpm, t_idx, sigma_here,
                        model_1d, model_2d, prior_info_2d,
                        switch_sigma=SWITCH_SIGMA, temporal_first=TEMPORAL_FIRST,
                        blend_h=BLEND_H, blend_band=BLEND_BAND):
    """Soft-blended eps prediction (replaces the old hard switch).

    Convex-blends the 2D spatial and 1D temporal eps predictions with the
    sigmoid weight `blend_weight_high`.  BOTH priors are evaluated only inside
    the crossover band [switch/band, switch*band]; outside it a single prior is
    used, so the cost matches the old hard switch except in the narrow band.

    Returns (eps[B,T,H,W] in DDPM convention, tag).
    """
    high_is_2d = not temporal_first
    lo, hi = switch_sigma / blend_band, switch_sigma * blend_band

    def call_2d():
        return eps_from_2d_prior(x_ddpm, t_idx, model_2d, prior_info_2d)
    def call_1d():
        return eps_from_1d_prior(x_ddpm, t_idx, model_1d)

    # --- outside the band: single prior (cheap path, identical cost to hard switch) ---
    if sigma_here >= hi:
        return (call_2d(), "2D") if high_is_2d else (call_1d(), "1D")
    if sigma_here <= lo:
        return (call_1d(), "1D") if high_is_2d else (call_2d(), "2D")

    # --- inside the band: convex blend of both priors ---
    w_high = blend_weight_high(sigma_here, switch_sigma, blend_h)
    eps2d, eps1d = call_2d(), call_1d()
    if high_is_2d:
        eps = w_high * eps2d + (1.0 - w_high) * eps1d
    else:
        eps = w_high * eps1d + (1.0 - w_high) * eps2d
    return eps, f"blend(w_high={w_high:.2f})"


@torch.no_grad()
def predict_eps_composed(x_ddpm, t_idx, sigma_here,
                         model_1d, model_2d, prior_info_2d,
                         mode=None,
                         switch_sigma=SWITCH_SIGMA, temporal_first=TEMPORAL_FIRST,
                         blend_h=BLEND_H, blend_band=BLEND_BAND):
    """Composition dispatcher for the two prior eps-predictions.

    mode="avg"   : eps = (eps_2d + eps_1d)/2  -- naive averaging (default for now).
                   Equivalent to averaging the scores: the geometric mean
                   p_S^{1/2} p_T^{1/2} (tempered product of experts). Both experts
                   act at EVERY sigma -- no band, no w->0 hand-off.
    mode="sum"   : eps = eps_2d + eps_1d      -- full product of experts; roughly
                   halve the Langevin step size to compensate the doubled score.
    mode="blend" : the old sigmoid soft blend (kept for A/B comparison).

    Returns (eps[B,T,H,W] in DDPM convention, tag).
    """
    mode = PRIOR_COMPOSITION if mode is None else mode
    if mode == "blend":
        return predict_eps_blended(x_ddpm, t_idx, sigma_here,
                                   model_1d, model_2d, prior_info_2d,
                                   switch_sigma, temporal_first, blend_h, blend_band)
    eps2d = eps_from_2d_prior(x_ddpm, t_idx, model_2d, prior_info_2d)
    eps1d = eps_from_1d_prior(x_ddpm, t_idx, model_1d)
    if mode == "avg":
        return 0.5 * (eps2d + eps1d), "avg"
    if mode == "sum":
        return eps2d + eps1d, "sum"
    raise ValueError(f"unknown PRIOR_COMPOSITION mode: {mode!r}")


@torch.no_grad()
def denoise_at_eps(x, t_eps, model_1d, model_2d, prior_info_2d,
                   sigma_arr, alphas_cumprod_t, mode=None):
    """Tweedie MMSE denoiser at the smoothing level eps:
        D_eps(x) = E[x_0 | x_0 + eps*eta = x]  under the composed prior.

    x lives in clean/EDM x0-space; the eps-networks (DDPM convention) see
    sqrt(abar_eps) * x at the small timestep t_eps, and
        D_eps(x) = x - sigma_eps * eps_hat(sqrt(abar_eps) x, t_eps).
    Used (a) as the refreshed prior ANCHOR inside the PnP-ULA loop (Sec.10-ii)
    and (b) averaged over the final chain as the MMSE readout (Sec.10-iii).
    """
    t_eps     = int(t_eps)
    sigma_eps = float(sigma_arr[t_eps])
    sqab_eps  = torch.sqrt(alphas_cumprod_t[t_eps])
    eps, _ = predict_eps_composed(sqab_eps * x, t_eps, sigma_eps,
                                  model_1d, model_2d, prior_info_2d, mode=mode)
    eps = torch.nan_to_num(eps, nan=0.0, posinf=0.0, neginf=0.0)
    return x - sigma_eps * eps


@torch.no_grad()
def prior_score_smoothed(x, t_eps, model_1d, model_2d, prior_info_2d,
                         sigma_arr, alphas_cumprod_t, mode=None):
    """PnP-ULA prior score s_{theta,eps}(x) = grad log p_eps(x)
       = (D_eps(x) - x) / sigma_eps^2  =  -eps_hat / sigma_eps   (Tweedie).
    Convenience wrapper around denoise_at_eps."""
    sigma_eps = float(sigma_arr[int(t_eps)])
    D = denoise_at_eps(x, t_eps, model_1d, model_2d, prior_info_2d,
                       sigma_arr, alphas_cumprod_t, mode=mode)
    return (D - x) / (sigma_eps ** 2)
````

## 8. Poisson log-likelihood gradient (analytical, voxel-wise)

From §1 of the Spatio-Temporal Guidance note, the Poisson log-likelihood is voxel-wise separable:
$$\log p(N\mid x)\;=\;\sum_{k,i,j}\big[N_{kij}\log\mu_{kij}-\mu_{kij}\big]+\text{const},\qquad \mu_{kij}=B_k(\alpha\phi_{kij}+d).$$
Differentiating through $\phi=\exp(\tfrac{L}{2}(x+1))$:
$$\nabla_x\log p(N\mid x)\;=\;\Big(\tfrac{N}{\mu}-1\Big)\,\underbrace{B\,\alpha\,\phi\,\tfrac{L}{2}}_{=\,\partial\mu/\partial x}.$$
Computed analytically (no autograd), elementwise, in normalized log-flux space.

**Pooled (photon-count-indexed) variant.**  `poisson_grad_pooled_uniform` sums photon counts and exposures over groups of $P$ consecutive bins (the Poisson sufficient statistic $\sum_k N_k$, exposure $\sum_k B_k\alpha$) and assumes a shared flux per group, so the data term constrains only the pooled mean — **injecting temporal coupling** — while the prior resolves within-pool detail.  $P=1$ recovers the exact per-bin score.  The DAPS loop (§11) anneals $P$ coarse$\to$fine from `POOL_FACTOR_INIT` to $1$: high effective photons-per-measurement at high noise (fewer salt-and-pepper spikes), full temporal resolution at low noise.

````python
def poisson_log_likelihood_grad(x_norm, bin_counts, bin_sizes, ppp_scale,
                                 dark_count=0.0, log_flux_max=LOG_FLUX_MAX,
                                 mu_eps=1e-8):
    """
    Analytical gradient of log p(N|x) for Poisson SPAD model, in normalized log-flux space.

    Args
    ----
    x_norm     : [B,T,H,W] normalized log-flux in [-1,1].
    bin_counts : [B,T,H,W] photon counts (or [T,H,W], will be unsqueezed).
    bin_sizes  : [T] or [B,T] frames per bin.
    ppp_scale  : scalar (alpha in the writeup).
    dark_count : scalar.
    log_flux_max : scalar L.

    Returns
    -------
    score [B,T,H,W] equal to (bc/mu - 1) * bs * ppp * phi * (L/2).
    """
    if bin_counts.dim() == 3:
        bin_counts = bin_counts.unsqueeze(0)
    if bin_sizes.dim() == 1:
        bs = bin_sizes.view(1, -1, 1, 1)
    elif bin_sizes.dim() == 2:
        bs = bin_sizes.view(bin_sizes.shape[0], bin_sizes.shape[1], 1, 1)
    else:
        bs = bin_sizes
    half_scale = log_flux_max / 2.0
    log_flux   = (x_norm + 1.0) * half_scale
    phi        = torch.exp(log_flux)
    bs_ppp_phi = bs * ppp_scale * phi
    mu         = bs_ppp_phi + bs * dark_count
    mu         = torch.clamp(mu, min=mu_eps)
    return (bin_counts / mu - 1.0) * bs_ppp_phi * half_scale


def poisson_grad_pooled_uniform(x_norm, bin_counts, bin_sizes, ppp_scale, pool_factor,
                                dark_count=0.0, log_flux_max=LOG_FLUX_MAX, mu_eps=1e-8):
    """
    Pooled Poisson score with TEMPORAL aggregation (Tier-A photon-count pooling).

    `pool_factor` consecutive bins are assumed to share one flux value, so the
    likelihood constrains only the pooled mean -- injecting coupling across time --
    and the diffusion prior resolves within-pool detail.  The pooled sufficient
    statistic is the per-pool total count Y_g = sum_k N_k, with pooled exposure
    E_g = sum_k B_k * alpha; the per-bin score is the pooled score broadcast back
    to every member bin.  pool_factor == 1 reproduces poisson_log_likelihood_grad
    exactly.  Requires T % pool_factor == 0.

    Args
    ----
    x_norm     : [B,T,H,W] normalized log-flux in [-1,1].
    bin_counts : [B,T,H,W] (or [T,H,W]) photon counts.
    bin_sizes  : [T] frames per bin (B_k).
    ppp_scale  : scalar (alpha).
    pool_factor: int temporal pooling factor P (>=1).

    Returns
    -------
    score [B,T,H,W].
    """
    if pool_factor is None or int(pool_factor) <= 1:
        return poisson_log_likelihood_grad(
            x_norm, bin_counts, bin_sizes, ppp_scale,
            dark_count=dark_count, log_flux_max=log_flux_max, mu_eps=mu_eps)

    if bin_counts.dim() == 3:
        bin_counts = bin_counts.unsqueeze(0)
    B, T, H, W = x_norm.shape
    P = int(pool_factor)
    assert T % P == 0, f"T={T} not divisible by pool_factor={P}"
    G = T // P
    half_scale = log_flux_max / 2.0

    Ek   = bin_sizes.view(1, T, 1, 1) * ppp_scale                       # exposure B_k*alpha per bin
    Yg   = bin_counts.view(B, G, P, H, W).sum(dim=2)                    # pooled counts   [B,G,H,W]
    Eg   = Ek.view(1, G, P, 1, 1).sum(dim=2)                            # pooled exposure [1,G,1,1]
    Dg   = (bin_sizes.view(1, G, P, 1, 1) * dark_count).sum(dim=2)      # pooled dark term [1,G,1,1]
    xbar = x_norm.view(B, G, P, H, W).mean(dim=2)                       # pooled-avg log-flux [B,G,H,W]

    phi    = torch.exp((xbar + 1.0) * half_scale)
    Eg_phi = Eg * phi
    mu     = torch.clamp(Eg_phi + Dg, min=mu_eps)                       # pooled rate
    score_g = (Yg / mu - 1.0) * Eg_phi * half_scale                     # [B,G,H,W]
    # Broadcast the pooled score back to every bin in the pool.
    return score_g.unsqueeze(2).expand(B, G, P, H, W).reshape(B, T, H, W)
````

## 9. Inner reverse PF-ODE (deterministic DDIM in EDM space)

Throughout DAPS we keep the iterate $x$ in **EDM convention** ($x_t=x_0+\sigma_t\varepsilon$), since this makes scores and Tweedie estimates clean.  When we need to call the ε-prediction networks (which were trained in DDPM convention), we form
$$x^{\text{DDPM}}_t=\sqrt{\bar\alpha_t}\,x^{\text{EDM}}_t.$$

The deterministic DDIM update in EDM convention reduces to a single line:
$$x^{\text{EDM}}_{t'}=x^{\text{EDM}}_{t}+(\sigma_{t'}-\sigma_t)\,\hat\varepsilon\big(x^{\text{DDPM}}_t,t\big),\qquad \sigma_{t'}<\sigma_t.$$
(Derivation: $\hat x_0=x^{\text{EDM}}_t-\sigma_t\hat\varepsilon$; $x^{\text{EDM}}_{t'}=\hat x_0+\sigma_{t'}\hat\varepsilon$.)

We take `DAPS_NUM_INNER_PFODE` such steps from $t_{\text{start}}$ to $t=0$.  In the repaired sampler this only provides the **warm start** for the PnP-ULA chain (§10) — the coupled prior work now happens inside the ULA itself.  At each substep the prior is composed per `PRIOR_COMPOSITION` (§7): naive averaging by default, so both experts act at every $\sigma$.

````python
@torch.no_grad()
def inner_pfode_ddim(x_edm, t_start, num_inner_steps,
                     model_1d, model_2d, prior_info_2d,
                     sigma_arr, alphas_cumprod_t,
                     prior_mode=None,
                     switch_sigma=SWITCH_SIGMA, temporal_first=TEMPORAL_FIRST,
                     blend_h=BLEND_H, blend_band=BLEND_BAND):
    """
    Deterministic DDIM reverse from t_start down to 0 in `num_inner_steps` substeps,
    using the composed prior (PRIOR_COMPOSITION; naive averaging by default) at every
    substep.  In the repaired sampler this only supplies the warm start for PnP-ULA.

    Args
    ----
    x_edm   : [B,T,H,W] iterate in EDM convention (x = x_0 + sigma * eps).
    t_start : int discrete timestep (0..NUM_DIFFUSION_STEPS-1) at which x_edm lives.

    Returns
    -------
    x0_hat  : [B,T,H,W] estimate of the clean signal.
    """
    if num_inner_steps <= 1:
        # Single Tweedie step (no inner ODE) -- still correct, just cheaper.
        ab    = alphas_cumprod_t[int(t_start)]
        sqab  = torch.sqrt(ab)
        sigma = sigma_arr[int(t_start)]
        x_ddpm = sqab * x_edm
        eps, _ = predict_eps_composed(x_ddpm, int(t_start), float(sigma),
                                      model_1d, model_2d, prior_info_2d, mode=prior_mode,
                                      switch_sigma=switch_sigma, temporal_first=temporal_first,
                                      blend_h=blend_h, blend_band=blend_band)
        return x_edm - sigma * eps

    t_seq = np.linspace(int(t_start), 0, num_inner_steps + 1)
    t_seq = np.unique(np.clip(t_seq.astype(np.int64), 0, len(sigma_arr) - 1))[::-1]
    x = x_edm
    for i in range(len(t_seq) - 1):
        t_cur  = int(t_seq[i])
        t_next = int(t_seq[i + 1])
        sigma_cur  = sigma_arr[t_cur]
        sigma_next = sigma_arr[t_next]
        ab_cur     = alphas_cumprod_t[t_cur]
        sqab_cur   = torch.sqrt(ab_cur)
        x_ddpm     = sqab_cur * x                                    # EDM -> DDPM
        eps, _     = predict_eps_composed(x_ddpm, t_cur, float(sigma_cur),
                                          model_1d, model_2d, prior_info_2d, mode=prior_mode,
                                          switch_sigma=switch_sigma, temporal_first=temporal_first,
                                          blend_h=blend_h, blend_band=blend_band)
        eps        = torch.nan_to_num(eps, nan=0.0, posinf=0.0, neginf=0.0)
        # DDIM in EDM:  x_{t'} = x_t + (sigma_{t'} - sigma_t) * eps.
        x = x + (sigma_next - sigma_cur) * eps
    return x
````

## 10. Repaired inner loop: annealed PnP-ULA over $p_\varepsilon(x_0\mid x_t,y)$

**Baseline DAPS (what we replace).** The anchor score $(\hat x_0-x)/\sigma_t^2$ collapses $p(x_0\mid x_t)$ to an isotropic Gaussian, making the inner target $p(y\mid x_0)\,\mathcal N(\hat x_0,\sigma_t^2 I)$ **voxel-separable**: its stationary law is a product of per-voxel marginals and photon evidence can never move sideways.

**Repaired chain.** Target the un-collapsed posterior (with an $\varepsilon$-smoothed prior):
$$\pi_t(x)\;\propto\;\underbrace{p(y\mid x)}_{\text{exact Poisson, separable by physics}}\;\cdot\;\underbrace{q(x_t\mid x)}_{\text{exact Gaussian tether}}\;\cdot\;\underbrace{p_\varepsilon(x)}_{\text{coupled denoiser prior}},$$
$$x^{(k+1)}=x^{(k)}+\delta\Big[\lambda_d\,\nabla\log p(N\mid x^{(k)})\;+\;\tfrac{x_t-x^{(k)}}{\sigma_t^2}\;+\;s_{\theta,\varepsilon}\big(x^{(k)}\big)\Big]+\sqrt{2\delta}\,\xi^{(k)},\qquad x^{(0)}=\hat x_0 .$$

The three discretization choices below are the analysis-backed ones — every constant comes from the model, none from a tuned schedule.

**(i) Step size from the target's own curvature.** ULA is stable with $O(\delta)$ bias iff $\delta\,L_\pi<1$, where $L_\pi$ bounds the Lipschitz constant of $\nabla\log\pi_t$ (Durmus & Moulines 2017; Laumont et al. 2022). Every term's constant is closed-form here:
$$L_\pi=\underbrace{\tfrac{1}{\sigma_t^2}}_{\text{tether (exact quadratic)}}+\underbrace{\tfrac{1}{\sigma_\varepsilon^2}}_{\nabla^2\log p_\varepsilon\succeq-\frac{I}{\varepsilon^2}}+\lambda_d\underbrace{\big(\tfrac{L}{2}\big)^2 E_{\max}e^{L}}_{L_y\text{ (data, exact for }d=0)},\qquad \boxed{\;\delta=\frac{c}{L_\pi},\;c=\texttt{PNP\_STEP\_SAFETY}\;}$$
For $d=0$ the log-flux link gives the Poisson score $(N-\mu(x))\tfrac{L}{2}$ with derivative $-(\tfrac{L}{2})^2\mu(x)$, bounded by $L_y$ on $x\in[-1,1]$ **independently of the counts** — this bound is exact, and pooling does not increase it (grouped exposure is redistributed over member bins). *Why the first run failed:* $\delta$ was still governed by the baseline's annealed lr ($10^{-3}\!\to\!10^{-5}$, designed for 40 free iterations × 100 steps), which bound the min at **every** step — integrated Langevin time fell 8× (data relaxation ≈ 0.5 e-folds vs ≈ 4.3 for baseline), so the likelihood never sculpted the iterate and the output was a prior-consensus sample. The fix is not a bigger lr but *deleting the schedule* in favour of the stability-bound step, which adapts to each annealing level automatically.

**(ii) Freeze the anchor, not the score (MYULA-style refreshes).** With the Tweedie denoiser $D_\varepsilon(x)=x+\sigma_\varepsilon^2 s_{\theta,\varepsilon}(x)$, the prior score is $s_{\theta,\varepsilon}(x)=\big(D_\varepsilon(x)-x\big)/\sigma_\varepsilon^2$. Between network refreshes (`PNP_SCORE_EVERY`) we freeze the **anchor** $\bar D=D_\varepsilon(x^{(k_0)})$ and use $(\bar D-x)/\sigma_\varepsilon^2$: the expensive slowly-varying part is lagged, the stiff linear part stays exact, so the restoring curvature $-1/\sigma_\varepsilon^2$ is present in every iteration — consistent with the step rule in (i). This is the Moreau–Yosida / proximal-anchor construction of MYULA (Durmus, Moulines & Pereyra 2018) with the proximal map replaced by the MMSE denoiser (Laumont et al. 2022); `PNP_SCORE_EVERY=1` recovers exact PnP-ULA. (Freezing the score *vector* — the previous implementation — leaves no restoring force between refreshes.)

**(iii) Posterior-mean readout; the sample is kept, not discarded.** $\pi_t$ at the final level is (approximately) the posterior convolved with $\mathcal N(0,\sigma_\varepsilon^2)$, so a *sample* legitimately carries equilibrium fluctuation $\approx L_\pi^{-1/2}\approx0.04$ in $x$ (≈ ±20% flux) — reporting a raw sample confounds honest posterior spread with the smoothing dither. Under the smoothing model $x=x_0+\varepsilon\eta$, the MMSE estimate of $x_0$ is $\mathbb E_{\pi}[\,\mathbb E(x_0\mid x)\,]=\mathbb E_\pi[D_\varepsilon(x)]$ — estimated by averaging the anchors from the post-burn-in half of the final chain (Rao–Blackwellized, zero extra network calls). This is a Monte-Carlo estimator of the conditional mean, not a post-hoc filter; the raw final sample is returned alongside it for posterior-spread reporting.

**Cost.** Per outer step: $2\cdot$`num_inner_pfode` $+\;2\lceil K/\texttt{PNP\_SCORE\_EVERY}\rceil$ full-volume network passes (both priors per evaluation under `"avg"`).

````python
@torch.no_grad()
def langevin_pnp_ula(x0_init, x_t, sigma_t,
                     bin_counts, bin_sizes, ppp_scale,
                     model_1d, model_2d, prior_info_2d,
                     sigma_arr, alphas_cumprod_t,
                     num_steps, t_eps,
                     prior_mode=None, score_every=1,
                     step_safety=PNP_STEP_SAFETY,
                     dark_count=0.0, log_flux_max=LOG_FLUX_MAX,
                     lambda_data=DAPS_LAMBDA_DATA, clip_x=DAPS_CLIP_X,
                     pool_factor=1, return_readout=False, verbose=False):
    """
    Annealed PnP-ULA targeting  pi_t(x) ~ p(y|x) q(x_t|x) p_eps(x).

    Discretization (see Sec.10 -- every constant comes from the model, none tuned):

    * Step size:  delta = step_safety / L_pi  with
          L_pi = 1/sigma_t^2 + 1/sigma_eps^2 + lambda_data * L_y,
      the sum of the closed-form Lipschitz constants of the three drift terms
      (ULA stability requires delta * L_pi < 1; Durmus & Moulines 2017,
      Laumont et al. 2022).  L_y = (L/2)^2 * E_max * e^L is EXACT for
      dark_count = 0: the log-link makes the Poisson score (N - mu(x)) * L/2
      with derivative -(L/2)^2 mu(x), bounded on x in [-1,1] independently of
      the observed counts; pooling does not increase the bound.  No learning-
      rate schedule.

    * Prior term: at refresh iterations evaluate the Tweedie anchor
      D_eps(x) and use s(x) = (D_eps_bar - x)/sigma_eps^2, keeping the ANCHOR
      frozen for `score_every` iterations.  The exact restoring curvature
      -1/sigma_eps^2 is retained every iteration (MYULA construction: Durmus,
      Moulines & Pereyra 2018, prox -> MMSE denoiser per Laumont et al.).
      score_every = 1 recovers exact PnP-ULA.

    * Readout (return_readout=True): average the anchors from the post-burn-in
      half of the chain -- the Rao-Blackwellized MCMC estimator of
      E_pi[D_eps(x)] = E[x_0 | .] under the eps-smoothing model, at zero extra
      network cost.  The raw final sample is returned alongside for
      posterior-spread reporting.

    Returns
    -------
    (x_sample, readout) : readout is None unless return_readout=True.
    """
    sigma_t_sq   = float(sigma_t) ** 2
    sigma_eps    = float(sigma_arr[int(t_eps)])
    sigma_eps_sq = sigma_eps ** 2
    half_scale   = log_flux_max / 2.0

    # Closed-form Lipschitz bound of the drift and the resulting stable step.
    E_max = float(torch.as_tensor(bin_sizes).max()) * float(ppp_scale)
    L_y   = (half_scale ** 2) * E_max * math.exp(log_flux_max)   # exact for dark_count == 0
    L_pi  = 1.0 / sigma_t_sq + 1.0 / sigma_eps_sq + float(lambda_data) * L_y
    delta = float(step_safety) / L_pi
    sqrt_2d = math.sqrt(2.0 * delta)
    score_every = max(1, int(score_every))

    x = x0_init.clone()
    anchor = None
    readout_sum, readout_n = None, 0
    for k in range(num_steps):
        if anchor is None or (k % score_every) == 0:
            anchor = denoise_at_eps(x, t_eps, model_1d, model_2d, prior_info_2d,
                                    sigma_arr, alphas_cumprod_t, mode=prior_mode)
            if return_readout and k >= num_steps // 2:
                readout_sum = anchor.clone() if readout_sum is None else readout_sum + anchor
                readout_n  += 1
        prior_score = (anchor - x) / sigma_eps_sq
        data_score  = poisson_grad_pooled_uniform(
            x, bin_counts, bin_sizes, ppp_scale, pool_factor,
            dark_count=dark_count, log_flux_max=log_flux_max,
        )
        tether = (x_t - x) / sigma_t_sq
        score  = lambda_data * data_score + tether + prior_score
        score  = torch.nan_to_num(score, nan=0.0, posinf=0.0, neginf=0.0)
        x = x + delta * score + sqrt_2d * torch.randn_like(x)
        if clip_x:
            x = x.clamp_(-1.0, 1.0)   # projection onto the training domain
        if verbose and (k % max(1, num_steps // 4) == 0):
            ds = data_score.norm().item()  / x.numel() ** 0.5
            ts = tether.norm().item()      / x.numel() ** 0.5
            ps = prior_score.norm().item() / x.numel() ** 0.5
            print(f"    [PnP-ULA] iter={k:3d}  delta={delta:.2e}  "
                  f"L_pi=({1.0/sigma_t_sq:.1e}+{1.0/sigma_eps_sq:.1e}+{lambda_data*L_y:.1e})  "
                  f"pool={pool_factor}  ||data||={ds:.3e}  ||tether||={ts:.3e}  ||prior||={ps:.3e}")

    readout = None
    if return_readout:
        if readout_n > 0:
            readout = readout_sum / readout_n
        else:   # chain too short for post-burn-in anchors: single Tweedie readout
            readout = denoise_at_eps(x, t_eps, model_1d, model_2d, prior_info_2d,
                                     sigma_arr, alphas_cumprod_t, mode=prior_mode)
    return x, readout
````

## 11. Repaired DAPS sampler (outer annealing loop)

Putting it together:

```
x_T  ~  N(0, sigma_max^2 I)                                            # init in EDM space
for step = 0,...,K-1:
    sigma_t        = sigma_arr[t_indices[step]]
    x0_hat         = inner_pfode_ddim(x_t, t_indices[step], num_inner_pfode)   # warm start
    x0_y, readout  = langevin_pnp_ula(x0_hat, x_t, sigma_t, y, t_eps,          # delta = c / L_pi
                                      return_readout = (step == K-1))
    x_t            = x0_y + sigma_next * eps,   eps ~ N(0,I)                   # exact forward renoising
return readout (MMSE estimate), x0_y (posterior sample), trajectory
```

Differences from the baseline loop: the ULA receives the noisy iterate $x_t$ (exact tether anchor) and the smoothing timestep $t_\varepsilon$; the collapsed DAPS anchor $\mathcal N(\hat x_0,\sigma_t^2 I)$ is gone; there is **no learning-rate schedule anywhere** — the step size is recomputed at every annealing level from the closed-form curvature $L_\pi$ of that level's target (§10-i). Prior composition is `PRIOR_COMPOSITION` everywhere (naive averaging by default). Photon-count pooling defaults to **off**.

**Two outputs, two roles.** `x0_mmse` is the Rao–Blackwellized posterior-mean readout from the final chain (§10-iii) — the correct object for SSIM/PSNR/LPIPS and visual comparison. `x0_sample` is an honest draw from the $\varepsilon$-smoothed posterior — the correct object for sample-variance / posterior-spread reporting (run the sampler several times for that). Their difference is itself informative: it is (an estimate of) the chain's equilibrium fluctuation $\approx L_\pi^{-1/2}$.

````python
def pool_factor_for_ratio(ratio, pool_init=PNP_POOL_INIT, vol_t=VOL_T):
    """Coarse->fine temporal pool factor over the annealing schedule.

    pool_init at ratio=0 (high noise) -> 1 at ratio=1 (low noise), snapped to a
    power of two that divides vol_t.  pool_init <= 1 disables pooling (returns 1).
    Pooling is OFF by default in the repaired sampler (PNP_POOL_INIT=1).
    """
    if pool_init is None or pool_init <= 1:
        return 1
    steps = int(math.floor(math.log2(pool_init)))          # largest power-of-2 pool <= pool_init
    k     = int(round((1.0 - float(ratio)) * steps))        # halvings applied at this ratio
    P     = 1 << max(0, min(steps, k))
    while P > 1 and (vol_t % P) != 0:                       # safety: keep P | vol_t
        P //= 2
    return P


def sample_daps_3d(
    bin_counts, bin_sizes, ppp_scale,
    model_1d, model_2d, prior_info_2d,
    sigma_arr, alphas_cumprod_t,
    vol_shape=(1, VOL_T, VOL_H, VOL_W),
    num_annealing=DAPS_NUM_ANNEALING,
    num_inner_pfode=DAPS_NUM_INNER_PFODE,
    num_mcmc=PNP_NUM_MCMC,
    lambda_data=DAPS_LAMBDA_DATA,
    prior_mode=PRIOR_COMPOSITION, t_eps=None,
    score_every=PNP_SCORE_EVERY, step_safety=PNP_STEP_SAFETY,
    pool_init=PNP_POOL_INIT,
    switch_sigma=SWITCH_SIGMA, temporal_first=TEMPORAL_FIRST,
    blend_h=BLEND_H, blend_band=BLEND_BAND,
    dark_count=DARK_COUNT, log_flux_max=LOG_FLUX_MAX,
    snapshot_every=10, verbose=True,
):
    """
    Repaired DAPS posterior sampler for the SPAD Poisson inverse problem.

    Inner loop = annealed PnP-ULA (Sec.10): exact Gaussian tether to the noisy
    iterate x_t plus the coupled Tweedie anchor at smoothing level eps (t_eps),
    refreshed inside the loop; step size delta = step_safety / L_pi from the
    closed-form curvature of the target (no lr schedule).  Prior composition =
    `prior_mode` everywhere ("avg" naive averaging | "sum" PoE | "blend" legacy).
    Photon-count pooling defaults to OFF (pool_init=1).

    Returns
    -------
    x0_mmse   : [B,T,H,W] MMSE readout E_pi[D_eps(x)] from the final chain --
                the posterior-MEAN estimate; use this for metrics.
    x0_sample : [B,T,H,W] final chain state -- an (eps-smoothed) posterior
                SAMPLE; use for posterior-spread reporting.
    trajectory: list of (step_idx, t_idx, phase, x0_y_cpu) sample snapshots.
    """
    if t_eps is None:
        t_eps = PNP_EPS_T
    sigma_max = float(sigma_arr[-1])
    # Initialise in EDM space at sigma_max (= prior of (x_0 + sigma_max * eps) under any data prior).
    x_t = torch.randn(vol_shape, device=device) * sigma_max

    # Annealing schedule: uniform sweep through discrete timesteps 999 -> 0.
    t_indices = np.linspace(NUM_DIFFUSION_STEPS - 1, 0, num_annealing + 1).astype(np.int64)

    iterator = (tqdm(range(num_annealing), desc=f"DAPS-3D repaired (PnP-ULA, prior={prior_mode})")
                if verbose else range(num_annealing))

    trajectory = []
    x0_y, x0_mmse = None, None
    for step in iterator:
        t_cur     = int(t_indices[step])
        sigma_t   = float(sigma_arr[t_cur])
        ratio     = step / max(1, num_annealing - 1)                    # in [0, 1]
        last_step = (step == num_annealing - 1)

        # 1. Inner reverse PF-ODE -> x0_hat: WARM START only (composed prior).
        x0_hat = inner_pfode_ddim(x_t, t_cur, num_inner_pfode,
                                   model_1d, model_2d, prior_info_2d,
                                   sigma_arr, alphas_cumprod_t,
                                   prior_mode=prior_mode,
                                   switch_sigma=switch_sigma,
                                   temporal_first=temporal_first,
                                   blend_h=blend_h, blend_band=blend_band).clamp_(-1.0, 1.0)

        # 2. PnP-ULA on p(y|x) q(x_t|x) p_eps(x); curvature-matched step (Sec.10).
        pool_now = pool_factor_for_ratio(ratio, pool_init)              # 1 unless pooling re-enabled
        x0_y, readout = langevin_pnp_ula(
            x0_hat, x_t, sigma_t,
            bin_counts, bin_sizes, ppp_scale,
            model_1d, model_2d, prior_info_2d,
            sigma_arr, alphas_cumprod_t,
            num_steps=num_mcmc, t_eps=t_eps,
            prior_mode=prior_mode, score_every=score_every, step_safety=step_safety,
            dark_count=dark_count, log_flux_max=log_flux_max,
            lambda_data=lambda_data, clip_x=DAPS_CLIP_X,
            pool_factor=pool_now, return_readout=last_step, verbose=False,
        )
        if last_step:
            x0_mmse = readout                                           # Rao-Blackwellized E_pi[D_eps(x)]

        # 3. Forward diffusion to next noise level (in EDM convention).
        if not last_step:
            t_next     = int(t_indices[step + 1])
            sigma_next = float(sigma_arr[t_next])
            x_t = x0_y + sigma_next * torch.randn_like(x0_y)
        else:
            x_t = x0_y

        phase = f"PnP-ULA[{prior_mode}]"
        if (step % snapshot_every == 0) or last_step:
            trajectory.append((step, t_cur, phase, x0_y.detach().cpu().clone()))
        if verbose and ((step % max(1, num_annealing // 10) == 0) or last_step):
            lf_mean_flux = np.exp(denormalize_log_flux(x0_y.mean()).item())
            print(f"  step {step:3d}  t={t_cur:4d}  sigma={sigma_t:.3f}  pool={pool_now:2d}  "
                  f"t_eps={int(t_eps)}  K/refresh={num_mcmc}/{score_every}  "
                  f"exp(mean log-flux)={lf_mean_flux:.1f}  "
                  f"x0_y[min,max]=[{x0_y.min():.3f},{x0_y.max():.3f}]")
    return x0_mmse, x0_y, trajectory

print("✓ Repaired DAPS-3D sampler defined (curvature-matched PnP-ULA + MMSE readout)")
````

## 12b. Attribution ablations (run manually after the main run)

The flat first run had two candidate causes that the fixes above address only partially: step-size starvation (fixed by §10-i) and a possible **flattening bias of the naive average itself** — the averaged eps runs through a full reverse trajectory at every outer step, compounding the noised-marginal composition bias (Du et al., ICML 2023), and the consensus of "1D likes constant traces" × "2D likes smooth frames" is flat-at-mean. Two runs attribute it:

* **(A) Prior-only** (`lambda_data=0`, same composition): samples the annealed composed *prior*. If this is still flat, the naive average itself is the flattener — the composition, not the data path, and the average should then live only in the ULA anchor (with `"blend"` or a single expert in the warm-start ODE).
* **(B) Blend composition, full posterior** (`prior_mode="blend"`): identical repaired dynamics, legacy composition — isolates composition from sampler dynamics.

| A result | B result | Reading |
|---|---|---|
| structured | – | composition fine; if main run still weak, raise `num_mcmc` / `num_annealing` |
| flat | structured | naive average flattens through the ODE — move the average out of the warm start |
| flat | flat | dynamics still under-powered — raise the budget before touching composition |

````python
RUN_ABLATIONS = False   # flip to True and re-run this cell (needs the main run's variables in scope)

if RUN_ABLATIONS:
    assert "x0_mmse_3d" in globals(), "run the main sampling cell (Sec.12) first"
    ABL = dict(num_annealing=25, num_inner_pfode=1, num_mcmc=PNP_NUM_MCMC,
               snapshot_every=5, verbose=True)
    common = dict(bin_counts=bin_counts_3d_t, bin_sizes=bin_sizes_3d_t, ppp_scale=ppp_scale_3d,
                  model_1d=model_1d, model_2d=model_2d, prior_info_2d=prior_info_2d,
                  sigma_arr=sigma_arr, alphas_cumprod_t=alphas_cumprod_t,
                  vol_shape=(1, VOL_T, VOL_H, VOL_W),
                  t_eps=PNP_EPS_T, score_every=PNP_SCORE_EVERY, step_safety=PNP_STEP_SAFETY,
                  pool_init=PNP_POOL_INIT, dark_count=DARK_COUNT, log_flux_max=LOG_FLUX_MAX)

    # (A) prior-only: does the annealed composed prior alone produce structure?
    xA_mmse, xA_samp, _ = sample_daps_3d(lambda_data=0.0,
                                         prior_mode=PRIOR_COMPOSITION, **ABL, **common)
    # (B) legacy blend composition under the SAME repaired dynamics.
    xB_mmse, xB_samp, _ = sample_daps_3d(lambda_data=DAPS_LAMBDA_DATA,
                                         prior_mode="blend", **ABL, **common)

    k = VOL_T // 2
    panels = [(x0_mmse_3d, f"main: {PRIOR_COMPOSITION}, lambda={DAPS_LAMBDA_DATA}"),
              (xA_mmse,    f"A: prior-only ({PRIOR_COMPOSITION}, lambda=0)"),
              (xB_mmse,    "B: blend, full posterior")]
    fig, ax = plt.subplots(1, 3, figsize=(13, 4))
    for a, (vol, ttl) in zip(ax, panels):
        a.imshow(np.exp(denormalize_log_flux(vol[0, k].detach().cpu().numpy())), cmap="gray")
        a.set_title(ttl, fontsize=10); a.axis("off")
    plt.suptitle("Flattening attribution: is the naive average the flattener?")
    plt.tight_layout(); plt.show()

    for name, vol in [("A prior-only", xA_mmse), ("B blend", xB_mmse)]:
        v = vol[0].detach().cpu().numpy()
        print(f"{name}: x-range=[{v.min():.3f},{v.max():.3f}]  spatial-std(mid frame)={v[k].std():.4f}")
````

## 14. Per-frame image-quality metrics (SSIM / PSNR / LPIPS)

Same protocol as the DiffPIR notebook so numbers can be compared cell-for-cell.

````python
from skimage.metrics import structural_similarity as sk_ssim
try:
    from skimage.metrics import peak_signal_noise_ratio as sk_psnr
except Exception:
    sk_psnr = None

def _psnr_np(x, y, data_range):
    mse = np.mean((x.astype(np.float64) - y.astype(np.float64)) ** 2)
    return float('inf') if mse == 0 else float(20 * np.log10(data_range) - 10 * np.log10(mse))

T_, H_, W_ = flux_vol_true.shape
ssim_pf = np.zeros(T_); psnr_pf = np.zeros(T_)
for t in range(T_):
    gt, es = flux_vol_true[t], flux_est_vol[t]
    scale = max(gt.max(), 1e-8)
    gt_n = np.clip(gt / scale, 0.0, 1.0).astype(np.float32)
    es_n = np.clip(es / scale, 0.0, 1.0).astype(np.float32)
    ssim_pf[t] = sk_ssim(gt_n, es_n, data_range=1.0)
    psnr_pf[t] = sk_psnr(gt_n, es_n, data_range=1.0) if sk_psnr is not None else _psnr_np(gt_n, es_n, 1.0)

print(f"SSIM mean={ssim_pf.mean():.4f}  std={ssim_pf.std():.4f}  range=[{ssim_pf.min():.4f}, {ssim_pf.max():.4f}]")
print(f"PSNR mean={psnr_pf.mean():.2f} dB  std={psnr_pf.std():.2f}  range=[{psnr_pf.min():.2f}, {psnr_pf.max():.2f}]")

lpips_pf = None
try:
    import lpips as _lpips_mod
    if "_lpips_net" not in globals():
        _lpips_net = _lpips_mod.LPIPS(net="alex").to(device).eval()
        globals()["_lpips_net"] = _lpips_net
    lpips_pf = np.zeros(T_); LPIPS_BATCH = 32
    with torch.no_grad():
        for start in range(0, T_, LPIPS_BATCH):
            end = min(start + LPIPS_BATCH, T_)
            gt_chunk, es_chunk = flux_vol_true[start:end], flux_est_vol[start:end]
            scale = np.maximum(gt_chunk.reshape(end - start, -1).max(axis=1), 1e-8)[:, None, None]
            gt_n = np.clip(gt_chunk / scale, 0.0, 1.0).astype(np.float32)
            es_n = np.clip(es_chunk / scale, 0.0, 1.0).astype(np.float32)
            gt_t = torch.from_numpy(gt_n).to(device).unsqueeze(1).repeat(1, 3, 1, 1) * 2 - 1
            es_t = torch.from_numpy(es_n).to(device).unsqueeze(1).repeat(1, 3, 1, 1) * 2 - 1
            lpips_pf[start:end] = _lpips_net(gt_t, es_t).view(-1).cpu().numpy()
    print(f"LPIPS mean={lpips_pf.mean():.4f}  std={lpips_pf.std():.4f}  range=[{lpips_pf.min():.4f}, {lpips_pf.max():.4f}]")
except ImportError:
    print("LPIPS skipped: pip install lpips to enable.")

ncols = 3 if lpips_pf is not None else 2
fig, axes = plt.subplots(1, ncols, figsize=(5 * ncols, 3.2))
axes[0].plot(ssim_pf, color="#2ca02c"); axes[0].set_title(f"SSIM (mean={ssim_pf.mean():.4f})")
axes[0].set_xlabel("frame"); axes[0].grid(alpha=0.3)
axes[1].plot(psnr_pf, color="#1f77b4"); axes[1].set_title(f"PSNR (mean={psnr_pf.mean():.2f} dB)")
axes[1].set_xlabel("frame"); axes[1].grid(alpha=0.3)
if lpips_pf is not None:
    axes[2].plot(lpips_pf, color="#d62728"); axes[2].set_title(f"LPIPS (mean={lpips_pf.mean():.4f})")
    axes[2].set_xlabel("frame"); axes[2].grid(alpha=0.3)
plt.tight_layout(); plt.show()
````
